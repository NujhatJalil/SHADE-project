## Section 4: Changing Runtime Environments {#section4}

### Runtime Types

Colab offers three types of runtimes:
- **CPU**: Default, free tier
- **GPU**: NVIDIA GPU acceleration (free tier limited)
- **TPU**: Tensor Processing Unit (for TensorFlow models)

### Checking Current Runtime


```python
# Check available resources
import psutil
import os

print("=== Runtime Information ===")
print(f"CPU count: {psutil.cpu_count()}")
print(f"RAM: {psutil.virtual_memory().total / (1024**3):.2f} GB")

# Check for GPU
try:
    import tensorflow as tf
    print(f"\nGPU Available: {tf.config.list_physical_devices('GPU')}")
    if tf.config.list_physical_devices('GPU'):
        print("GPU Name:", tf.config.list_physical_devices('GPU')[0].name)
except:
    print("\nGPU: Not available or TensorFlow not installed")

# Check for TPU
try:
    import tensorflow as tf
    tpu = tf.distribute.cluster_resolver.TPUClusterResolver()
    print(f"\nTPU Available: {tpu.cluster_spec().as_dict()}")
except:
    print("\nTPU: Not available")
```

### How to Switch Runtimes

1. Click **Runtime** in the menu bar
2. Select **Change runtime type**
3. Choose:
   - **Hardware accelerator**: None (CPU), GPU, or TPU
   - **Runtime shape**: Standard or High-RAM (if available)

**Note**: Changing runtime will reset your notebook's execution state. You'll need to re-run all cells.

### Verifying GPU Setup


```python
# Verify GPU is available and working
import tensorflow as tf

print("TensorFlow version:", tf.__version__)
print("\nGPU Available:", tf.config.list_physical_devices('GPU'))

if tf.config.list_physical_devices('GPU'):
    print("\n✓ GPU is set up correctly!")
    # Test GPU computation
    with tf.device('/GPU:0'):
        a = tf.constant([[1.0, 2.0], [3.0, 4.0]])
        b = tf.constant([[1.0, 1.0], [0.0, 1.0]])
        c = tf.matmul(a, b)
    print(f"\nGPU computation test: {c.numpy()}")
else:
    print("\n✗ GPU not available. Make sure you've selected GPU runtime.")
```

## Section 7: Basic Notebook Operations {#section7}

### Running Cells

- **Run current cell**: `Ctrl+Enter` (or click the play button)
- **Run cell and move to next**: `Shift+Enter`
- **Run cell and insert new below**: `Alt+Enter`
- **Run all cells**: **Runtime → Run all**
- **Run all cells above**: **Runtime → Run before**
- **Run all cells below**: **Runtime → Run after**

### Cell Types

- **Code cell**: Executes Python code
- **Markdown cell**: Displays formatted text

Switch cell type: Click in the cell → **Code** or **Text** button, or use:
- `Ctrl+M M` for Markdown
- `Ctrl+M Y` for Code


```python
# Example: This is a code cell
print("Hello from Colab!")
print("This cell can execute Python code")

# You can use any Python library
import datetime
print(f"\nCurrent date: {datetime.datetime.now()}")
```

### Keyboard Shortcuts

Access shortcuts: **Tools → Keyboard shortcuts** or `Ctrl+M H`

**Common shortcuts:**
- `Ctrl+M A`: Insert cell above
- `Ctrl+M B`: Insert cell below
- `Ctrl+M D`: Delete cell
- `Ctrl+M Z`: Undo cell deletion
- `Ctrl+M X`: Cut cell
- `Ctrl+M C`: Copy cell
- `Ctrl+M V`: Paste cell
- `Ctrl+M M`: Convert to Markdown
- `Ctrl+M Y`: Convert to Code

### Downloading Outputs

You can download files, images, and data from Colab.


```python
# Download a file
from google.colab import files

# Create a sample file
with open('output.txt', 'w') as f:
    f.write("This is a file created in Colab\n")
    f.write("You can download it using files.download()\n")

# Download the file
files.download('output.txt')

# Download a CSV file
import pandas as pd
df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
df.to_csv('sample.csv', index=False)
files.download('sample.csv')

print("Files downloaded!")
```
